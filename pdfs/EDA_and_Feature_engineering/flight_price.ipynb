```python
def convert_into_minute(duration):
    hours,minutes = 0,0
    if 'h' in duration:
        hours = int(duration.split('h')[0].strip())
        duration = duration.split('h')[1].strip() if 'm' in duration else '0m'
    if 'm' in duration:
        minutes = int(duration.split('m')[0].strip())
    return (hours*60)+minutes
```

```python
df['Duration'] = df['Duration'].apply(convert_into_minute)
```
