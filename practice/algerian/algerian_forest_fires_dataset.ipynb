```python
df_cleaned = df.drop(index=122).reset_index(drop=True)
```

```python
df_cleaned.columns = df_cleaned.columns.str.strip()
```

```python
df_cleaned.columns
```

```python
df_cleaned.info()
```

```python
df_cleaned.head()
```

```python
df_formatted = df_cleaned.copy()
```

```python
df_formatted[['day','month','year','Temperature','RH','Ws']] = df_cleaned[['day','month','year','Temperature','RH','Ws']].astype(int)
```

```python
df_formatted.info()
```

```python
objects = [features for features in df_formatted.columns if df_formatted[features].dtype=='O']
```

```python
for i in objects:
    if i != 'Classes':
        df_formatted[i] = df_formatted[i].astype(float)
```

## Exploratory Data Analysis

```python
df_copy = df_formatted.drop(['day','month','year'],axis=1)
```

```python
df_copy.head()
```

### Encoding of the categories in classes

```python
df_copy['Classes'].value_counts()
```

```python
df_copy['Classes'] = np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy['Classes'].value_counts()
```

```python
plt.style.available
```

```python
# plot the density plot for all feature
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
#  plot piechart
classlabels = ['Fire','Not Fire']
values = df_copy['Classes'].value_counts(normalize=True)*100
plt.pie(values,labels=classlabels,autopct="%1.1f%%")
plt.title("Pie Chart of Classes")
plt.show()
```
